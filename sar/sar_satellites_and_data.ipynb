# SAR Satellites and Ocean Data Products

## SAR Satellites

Spaceborne SAR began with early experimental missions such as Seasat and has expanded into a major source of ocean and coastal information. Modern SAR missions provide repeated high-resolution microwave observations that are useful for winds, waves, sea ice, oil spills, ship detection, coastal flooding, and many other applications.

SAR data are often more complex to use than optical or infrared data because product choices depend on acquisition mode, polarization, incidence angle, orbit direction, processing level, and calibration status.

## Common SAR Missions for Oceanography

### Sentinel-1

Sentinel-1 is a C-band SAR mission in the European Copernicus program. It is one of the most widely used SAR sources for ocean and ice applications because of its systematic acquisitions, open data policy, and ocean-focused Level-2 products.

Sentinel-1 products support applications such as sea ice mapping, surface wind and wave information, oil spill detection, and maritime monitoring.

### RADARSAT Constellation Mission

The RADARSAT Constellation Mission is a Canadian C-band SAR mission. It supports maritime surveillance, sea ice monitoring, oil pollution monitoring, disaster response, and environmental applications.

### TerraSAR-X and PAZ

TerraSAR-X and PAZ are X-band SAR missions that can provide high-resolution imagery. X-band SAR is especially sensitive to small-scale roughness and can be useful for detailed coastal, ship, and surface-feature studies.

### ALOS-2 and ALOS-4 PALSAR

ALOS-2 and ALOS-4 carry L-band SAR systems. L-band has a longer wavelength than C-band and X-band, which can be useful for some ice, coastal wetland, and land-ocean boundary applications.

### SAOCOM

SAOCOM is an L-band SAR mission developed by Argentina. Its longer wavelength complements shorter-wavelength SAR missions for some environmental applications.

### NISAR

NISAR is a NASA-ISRO SAR mission with L-band and S-band radar. Its primary mission emphasis is global land and ice change, but its data are also relevant to coastal, cryosphere, wetland, and some ocean-adjacent applications.

## SAR Product Levels

SAR product names vary by mission, but common product types include:

| Product type | General meaning | Typical use |
|---|---|---|
| SLC | Single Look Complex data with amplitude and phase | advanced processing; preserves phase information |
| GRD | Ground Range Detected amplitude imagery | visual interpretation, calibrated backscatter, many ocean applications |
| Level-2 Ocean | derived geophysical ocean variables | winds, waves, radial velocity, or related ocean parameters depending on mission |

For this lesson, we focus on amplitude/backscatter and geophysical ocean products. Phase-based deformation processing is left for a separate lesson.

## Example: Comparing SAR Acquisition Choices

A single SAR mission can acquire data in different modes and polarizations. Product suitability depends on the oceanographic question.

For example:

- Wide-swath modes are useful for sea ice, winds, and maritime monitoring.
- Higher-resolution modes are useful for ships, coastlines, and small slicks.
- Co-polarized channels such as VV or HH are commonly used for wind and wave applications.
- Cross-polarized channels such as VH or HV can help with some high-wind, ice, and target detection applications.

```python
# A simple visual comparison of hypothetical spatial resolution and swath width
modes = ['Wide swath', 'Standard', 'High resolution']
resolution_m = np.array([100, 25, 5])
swath_km = np.array([400, 250, 30])

plt.figure(figsize=(7, 4))
plt.scatter(swath_km, resolution_m, s=120)
for mode, x, y in zip(modes, swath_km, resolution_m):
    plt.text(x + 5, y, mode, va='center')
plt.gca().invert_yaxis()
plt.xlabel('Approximate swath width (km)')
plt.ylabel('Approximate pixel size or resolution (m)')
plt.title('SAR Mode Trade-Off: Coverage vs Spatial Detail')
plt.grid(linestyle='--', linewidth=0.5, alpha=0.5)
plt.show()
```

## Ocean Data Products

### Sentinel-1 Level-2 Ocean Products

Sentinel-1 ocean products can include information related to wind, waves, and radial surface velocity depending on acquisition mode and processing. These products are useful when a derived geophysical variable is needed rather than a raw or calibrated SAR image.

### NOAA SAR Winds

NOAA distributes high-resolution sea surface wind products derived from SAR, including products based on RADARSAT Constellation Mission data. These products are designed for operational marine and coastal applications.

### Sea Ice Charts and Ice Products

National ice services use SAR imagery because it can observe polar regions through cloud cover and darkness. SAR supports ice edge mapping, ice concentration, ice type interpretation, and detection of deformation features.

### Oil Spill and Slick Mapping

SAR is often used for marine oil spill response because oil films can damp short surface waves and create dark features. Interpretation generally requires wind information and analyst judgment because natural slicks and low-wind regions can look similar.

### Ship Detection Products

SAR can detect ships as bright targets over darker ocean backgrounds. Detection can support maritime domain awareness, fisheries monitoring, and search-and-rescue applications.

## Practical Workflow for SAR Oceanography

A typical workflow is:

1. Define the oceanographic question and spatial scale.
2. Select a mission, acquisition mode, polarization, and product level.
3. Download calibrated or analysis-ready data when available.
4. Apply radiometric calibration, noise correction, terrain/ocean projection, and speckle reduction as needed.
5. Interpret backscatter patterns using wind, waves, currents, ice charts, ship records, or other context.
6. Validate with in situ data, models, or independent satellite observations when possible.

## Summary

SAR satellites provide active microwave observations that complement passive ocean color and infrared SST data.

Key ideas:

1. Sentinel-1, RADARSAT Constellation, TerraSAR-X/PAZ, ALOS PALSAR, SAOCOM, and NISAR are examples of SAR missions relevant to ocean or coastal studies.
2. Product choice matters: SLC, GRD, and Level-2 Ocean products serve different purposes.
3. Ocean SAR applications include winds, waves, slicks, ships, sea ice, fronts, and coastal hazards.
4. SAR can observe at night and through most clouds, making it especially valuable in storms and polar regions.
5. Interpretation requires attention to wavelength, polarization, incidence angle, acquisition mode, and environmental context.

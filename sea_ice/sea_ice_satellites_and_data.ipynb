## Satellites Used for Passive Microwave Sea Ice Observations

The passive microwave sea ice record extends back to the early 1970s. Important missions and sensors include:

| Mission / Program | Approximate period | Sensor or role | Notes |
|---|---:|---|---|
| Nimbus 5 | 1972-1983 | ESMR | Early passive microwave sea ice measurements near 19 GHz |
| Nimbus 7 | 1978-1994 | SMMR | Multi-channel passive microwave observations |
| DMSP-F8, F11, F13, F17 | 1987-present record continuity | SSM/I and SSMIS family | Long-running operational record used for sea ice concentration |
| Aqua | 2002-present | AMSR-E, then follow-on products from related AMSR sensors | Higher-frequency passive microwave observations alongside MODIS on Aqua |

The main frequencies used for sea ice concentration estimates include approximately 19, 22, and 37 GHz. Frequencies near 85/89/91 GHz provide additional information but are more susceptible to weather contamination.

```python
missions = pd.DataFrame({
    'mission': ['Nimbus 5', 'Nimbus 7', 'DMSP-F8', 'DMSP-F11', 'DMSP-F13', 'DMSP-F17', 'Aqua AMSR-E'],
    'start': [1972, 1978, 1987, 1991, 1995, 2006, 2002],
    'end':   [1983, 1994, 2006, 2000, 2015, 2026, 2011]
})
missions
```

```python
plt.figure(figsize=(8, 4))
for i, row in missions.iterrows():
    plt.plot([row['start'], row['end']], [i, i], linewidth=6)
plt.yticks(range(len(missions)), missions['mission'])
plt.xlabel('Year')
plt.title('Selected Passive Microwave Sea Ice Missions')
plt.xlim(1970, 2028)
plt.grid(axis='x', linestyle='--', linewidth=0.5)
plt.show()
```

## 10. Why Not Use Salinity-Frequency Radiometers for Sea Ice?

Sea surface salinity radiometers use microwave emission near 1.413 GHz. That frequency is very useful for salinity because the ocean microwave signal is sensitive to salinity at L-band.

For sea ice concentration, however, many historical passive microwave sensors did not collect data at 1.413 GHz. In addition, L-band observations require large antennas and careful correction for external noise sources. The long sea ice climate record instead relies primarily on higher microwave frequencies used by instruments such as ESMR, SMMR, SSM/I, SSMIS, AMSR-E, and related sensors.

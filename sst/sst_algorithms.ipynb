# SST Algorithms

## Motivation

The biggest challenge in satellite SST retrieval is that the atmosphere is also emitting and absorbing thermal infrared radiation. The satellite measures radiation at the top of the atmosphere, not just radiation emitted by the ocean surface.

A simplified view of the measured signal includes:

- thermal emission from the ocean surface
- absorption by atmospheric gases, especially water vapor
- thermal emission from the atmosphere
- small contributions from reflection, especially near 4 µm during daytime

SST algorithms use measurements from multiple infrared bands to estimate the surface temperature while correcting for atmospheric effects.

## Brightness Temperature

Instead of working directly with measured radiance, SST algorithms often use **brightness temperature**. Brightness temperature is the temperature a blackbody would need to have to emit the observed radiance at a particular wavelength or band.

A brightness temperature is not always equal to the true surface temperature because the atmosphere modifies the radiance before it reaches the sensor.

## The Split-Window Idea

Thermal infrared bands near 10-12 µm are affected differently by atmospheric water vapor. By comparing brightness temperatures from two nearby bands, algorithms can estimate and reduce atmospheric contamination.

This is the basis of the **multi-channel SST algorithm**.

## Multi-Channel SST Algorithm

A simplified multi-channel SST algorithm has the form:

$$SST = aT_1 + b(T_1 - T_2) + C$$

where:

- $T_1$ is the brightness temperature from one thermal infrared band
- $T_2$ is the brightness temperature from another thermal infrared band
- $T_1 - T_2$ helps account for atmospheric absorption and emission
- $a$, $b$, and $C$ are empirically calibrated constants

The exact coefficients depend on the sensor, viewing geometry, atmospheric conditions, and calibration data.

```python
# A toy example of a split-window SST retrieval.
# These coefficients are illustrative only and should not be used for real SST retrievals.

def simple_mcsst(T1, T2, a=1.0, b=2.5, C=0.0):
    """Illustrative multi-channel SST algorithm using two brightness temperatures."""
    return a*T1 + b*(T1 - T2) + C

T1 = 289.5  # brightness temperature from band 1, K
T2 = 288.9  # brightness temperature from band 2, K

sst_K = simple_mcsst(T1, T2)
sst_C = sst_K - 273.15

print(f'Band 1 brightness temperature: {T1:.2f} K')
print(f'Band 2 brightness temperature: {T2:.2f} K')
print(f'Illustrative SST estimate: {sst_K:.2f} K = {sst_C:.2f} °C')
```

## Sensor Bands Used for SST Retrievals

Different sensors use similar thermal infrared windows for SST retrieval. Examples include:

| Sensor | Band near 10-11 µm | Band near 11-12 µm |
|---|---:|---:|
| AVHRR | Band 4, 10.3-11.3 µm | Band 5, 11.5-12.5 µm |
| MODIS | Band 31, 10.78-11.28 µm | Band 32, 11.77-12.27 µm |
| Landsat TIRS | Band 10, 10.6-11.19 µm | Band 11, 11.5-12.51 µm |

The exact band placement matters because atmospheric absorption, surface emission, and instrument calibration differ by wavelength.

## Nonlinear SST Algorithms

More advanced algorithms extend the split-window approach by adding nonlinear terms, satellite viewing angle, and a first-guess or climatological SST value. These corrections help account for the changing path length through the atmosphere and the fact that atmospheric effects are not always linear.

A general nonlinear SST algorithm often includes terms related to:

- brightness temperature in one band
- the difference between two brightness temperatures
- viewing angle
- climatological or first-guess SST
- empirically calibrated coefficients

## Calibration with In Situ Observations

Satellite SST algorithms are continuously calibrated and validated against in situ observations. These include:

- Argo floats
- moored buoy arrays
- drifting buoys
- coastal and open-ocean observing platforms

This calibration is essential because satellite instruments measure radiance, while oceanographers usually want temperature.

## Argo Floats and Mooring Arrays

Argo floats measure temperature and salinity profiles throughout the global ocean. A typical Argo float drifts at depth, descends and ascends through the water column during a profile, then transmits data by satellite when it reaches the surface.

Moored arrays, such as the Tropical Atmosphere-Ocean (TAO) array, provide continuous oceanic and atmospheric measurements in key regions. These observations help validate satellite SST products and improve retrieval algorithms.

# Summary

Satellite SST retrievals rely on the relationship between temperature and thermal infrared emission. Because infrared radiation comes from only the ocean skin and must pass through the atmosphere, satellite SST is not a simple direct measurement of bulk upper-ocean temperature.

Key ideas:

1. Ocean-like temperatures emit peak radiation in the thermal infrared.
2. Infrared SST retrievals measure the ocean skin, not the full upper mixed layer.
3. Atmospheric windows near 4 µm and 11 µm are especially important for SST.
4. Multi-band algorithms help correct for atmospheric effects.
5. Higher-level products such as MUR and SST CCI are useful but should be interpreted carefully because they include merging and interpolation.
